# 2. b. Compare the mean and median of daily social media hours and sleep hours. What does the difference indicate about the distribution and skewness of these variables? 

```python
plt.figure(figsize=(10, 6))
plt.plot(df["daily_social_media_hours"])
plt.plot(df['sleep_hours'])
```

```python
sns.kdeplot(df["daily_social_media_hours"], color="crimson", label="Social Media", fill=True, alpha=0.05)
sns.kdeplot(df['sleep_hours'], color="royalblue", label="Sleep", fill=True, alpha=0.05)

# 3. Add Lines for Social Media
plt.axvline(df["daily_social_media_hours"].mean(), color="crimson", linestyle="--", linewidth=2, label="Social Media Mean")
plt.axvline(df["daily_social_media_hours"].median(), color="crimson", linestyle=":", linewidth=2, label="Social Media Median")

# 4. Add Lines for Sleep
plt.axvline(df["sleep_hours"].mean(), color="royalblue", linestyle="--", linewidth=2, label="Sleep Mean")
plt.axvline(df["sleep_hours"].median(), color="royalblue", linestyle=":", linewidth=2, label="Sleep Median")

# 5. Format chart
plt.title("Distribution with Mean and Median", fontsize=14)
plt.xlabel("Hours")
plt.ylabel("Density")
plt.legend() # This automatically updates to show the line styles
plt.show()
```

# 3. a. Identify all variables containing missing values. Calculate the number and percentage of missing values in each column.
b. Suggest suitable imputation techniques for missing values in daily social media hours, sleep hours, and screen time before sleep.
c. Suggest suitable imputation techniques for missing values in stress level, anxiety level, and addiction level.
d. Explain how missingness in mental health indicators such as anxiety level could potentially be MAR or MNAR. 


```python
def cal_missing(column_list):
    # Loop through each individual column in your list
    for column in column_list:
        missing_count = df[column].isnull().sum()
        missing_percentage = (missing_count / len(df)) * 100
        
        print(f"Column Name: {column}")
        print(f"\t Missing Values: {missing_count}")
        print(f"\t Percentage:     {missing_percentage:.2f}%\n")
```

```python
col_list = ['age', 'gender', 'daily_social_media_hours', 'platform_usage',
       'sleep_hours', 'screen_time_before_sleep', 'academic_performance',
       'physical_activity', 'social_interaction_level', 'stress_level',
       'anxiety_level', 'addiction_level', 'depression_label']

cal_missing(col_list)
```
